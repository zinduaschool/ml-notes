### Filtering in Pandas

```python
import pandas as pd
import numpy as np
```

```python
# load the dataset
athlete = pd.read_csv('bios.csv')
```

```python
athlete.head()
```

```python
athlete.info()
```

```python
# athlete.loc[athlete['height_cm'] > 200]
athlete[athlete['height_cm'] > 200] # shortform
```

```python
# athlete.loc[athlete['height_cm'] > 200, ['name', 'height_cm']] 

# use shortform to select columns

athlete[athlete['height_cm'] > 200] [['name', 'height_cm', 'born_country']]
```

```python
# combining conditions
athlete[(athlete['height_cm'] > 210) & (athlete['born_country'] =='USA')]
```

```python
# string operations
# athlete[athlete['name'].str.contains("Smith")]
# athlete[athlete['name'].str.contains("Smith", na=False)] # to avoid NaN error
athlete[athlete['name'].str.contains("smith", case=False)] # case insensitive 
```

```python
#isin 
# athlete[athlete['born_country'].isin(['USA', 'CAN'])] # filter by multiple values
athlete[athlete['born_country'].isin(['USA', 'CAN']) & (athlete['height_cm'] > 200)] # combine with other conditions
```

```python
athlete[athlete['born_country'].isin(['USA', 'CAN']) & (athlete['name'].str.startswith('James'))]
# athlete[athlete['born_country'].str.startswith('KEN')]
```

```python
# query function
athlete.query('born_country == "USA"')
```

```python
# athlete.query('born_country == "USA" & height_cm > 200') # combine with other conditions 

# athlete.query('born_country in ["USA", "CAN"]') # filter by multiple values

athlete.query('born_country in ["USA", "CAN"] & name.str.startswith("James")') # combine with other conditions
```
